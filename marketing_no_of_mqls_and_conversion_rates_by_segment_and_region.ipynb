```python
import pandas as pd

# ---------------------------------------------------------
# 1. OCR Data Extraction from Image 1 (Data Table)
# ---------------------------------------------------------
# Extracted manually to ensure 100% accuracy vs automated OCR
source_data = {
    "Request Contact": {"MQLs": 13534, "CVR_S2": "15%"},
    "Trial":           {"MQLs": 15078, "CVR_S2": "4%"},
    "Content":         {"MQLs": 2838,  "CVR_S2": "1%"},
    "Digital Event":   {"MQLs": 10657, "CVR_S2": "1%"},
    "Physical Event":  {"MQLs": 6714,  "CVR_S2": "2%"},
    "Grand Total":     {"MQLs": 48821, "CVR_S2": "6%"}
}

# ---------------------------------------------------------
# 2. OCR Data Extraction from Image 2 (Metric Labels)
# ---------------------------------------------------------
target_metrics = [
    "Marketing: Conversion Rate ( Content > S2 )",
    "Marketing: Conversion Rate ( Digital Event > S2 )",
    "Marketing: Conversion Rate ( Physical Event > S2 )",
    "Marketing: Conversion Rate ( Request > S2 )",
    "Marketing: Conversion Rate ( Trial > S2 )",
    "Marketing: MQL Volume",
    "Marketing: MQL volume ( Contact )",
    "Marketing: MQL volume ( Content )",
    "Marketing: MQL volume ( Events )",
    "Marketing: MQL volume ( Trials )"
]

# ---------------------------------------------------------
# 3. Mapping Logic (Creating the Summary Table)
# ---------------------------------------------------------
summary_data = []

for metric in target_metrics:
    value = "N/A" # Default

    # CVR Mappings
    if "Content > S2" in metric:
        value = source_data["Content"]["CVR_S2"]
    elif "Digital Event > S2" in metric:
        value = source_data["Digital Event"]["CVR_S2"]
    elif "Physical Event > S2" in metric:
        value = source_data["Physical Event"]["CVR_S2"]
    elif "Request > S2" in metric:
        value = source_data["Request Contact"]["CVR_S2"]
    elif "Trial > S2" in metric:
        value = source_data["Trial"]["CVR_S2"]

    # Volume Mappings
    elif "Marketing: MQL Volume" == metric: # Exact match for Grand Total
        value = source_data["Grand Total"]["MQLs"]
    elif "( Contact )" in metric:
        value = source_data["Request Contact"]["MQLs"]
    elif "( Content )" in metric:
        value = source_data["Content"]["MQLs"]
    elif "( Trials )" in metric:
        value = source_data["Trial"]["MQLs"]
    elif "( Events )" in metric:
        # Summing Digital + Physical for total "Events" volume
        value = source_data["Digital Event"]["MQLs"] + source_data["Physical Event"]["MQLs"]

    summary_data.append({"Metric Name": metric, "Value": value})

# Create DataFrame
df_summary = pd.DataFrame(summary_data)

# Display the table
df_summary
```

```python
import pandas as pd

# ---------------------------------------------------------
# 1. OCR Data Extraction
# ---------------------------------------------------------

# APAC (6.1k MQLs)
Apac_Ocr = {
    "Request Contact": {"MQLs": 737, "CVR_S2": "17%"},
    "Trial":           {"MQLs": 728, "CVR_S2": "4%"},
    "Content":         {"MQLs": 229, "CVR_S2": "0%"},
    "Digital Event":   {"MQLs": 1302, "CVR_S2": "1%"},
    "Physical Event":  {"MQLs": 3089, "CVR_S2": "1%"},
    "Grand Total":     {"MQLs": 6085, "CVR_S2": "3%"}
}

# EMEA (4.1k MQLs)
Emea_Ocr = {
    "Request Contact": {"MQLs": 830, "CVR_S2": "18%"},
    "Trial":           {"MQLs": 893, "CVR_S2": "5%"},
    "Content":         {"MQLs": 268, "CVR_S2": "1%"},
    "Digital Event":   {"MQLs": 962, "CVR_S2": "2%"},
    "Physical Event":  {"MQLs": 1135, "CVR_S2": "3%"},
    "Grand Total":     {"MQLs": 4088, "CVR_S2": "6%"}
}

# Digital (19.6k MQLs)
Dig_Ocr = {
    "Request Contact": {"MQLs": 7633, "CVR_S2": "12%"},
    "Trial":           {"MQLs": 9591, "CVR_S2": "3%"},
    "Content":         {"MQLs": 558, "CVR_S2": "0%"},
    "Digital Event":   {"MQLs": 1324, "CVR_S2": "0%"},
    "Physical Event":  {"MQLs": 543, "CVR_S2": "1%"},
    "Grand Total":     {"MQLs": 19649, "CVR_S2": "6%"}
}

# SMB (9.3k MQLs)
Smb_Ocr = {
    "Request Contact": {"MQLs": 2979, "CVR_S2": "20%"},
    "Trial":           {"MQLs": 2490, "CVR_S2": "8%"},
    "Content":         {"MQLs": 500, "CVR_S2": "0%"},
    "Digital Event":   {"MQLs": 2492, "CVR_S2": "3%"},
    "Physical Event":  {"MQLs": 863, "CVR_S2": "3%"},
    "Grand Total":     {"MQLs": 9324, "CVR_S2": "9%"}
}

# LATAM (2.1k MQLs)
Latam_Ocr = {
    "Request Contact": {"MQLs": 541, "CVR_S2": "12%"},
    "Trial":           {"MQLs": 474, "CVR_S2": "5%"},
    "Content":         {"MQLs": 167, "CVR_S2": "1%"},
    "Digital Event":   {"MQLs": 640, "CVR_S2": "0%"},
    "Physical Event":  {"MQLs": 314, "CVR_S2": "2%"},
    "Grand Total":     {"MQLs": 2136, "CVR_S2": "5%"}
}

# NA (formerly AMER - 7.5k MQLs)
Amer_Ocr = {
    "Request Contact": {"MQLs": 817, "CVR_S2": "24%"},
    "Trial":           {"MQLs": 924, "CVR_S2": "4%"},
    "Content":         {"MQLs": 1117, "CVR_S2": "1%"},
    "Digital Event":   {"MQLs": 3939, "CVR_S2": "1%"},
    "Physical Event":  {"MQLs": 772, "CVR_S2": "2%"},
    "Grand Total":     {"MQLs": 7569, "CVR_S2": "4%"}
}

# ---------------------------------------------------------
# 2. Define Function with Exact Metric Strings
# ---------------------------------------------------------

def process_metrics(source_data, column_name):
    # The exact string list provided by user
    raw_metrics_string = "Marketing: Conversion Rate ( Content > S2 ),Marketing: Conversion Rate ( Digital Event > S2 ),Marketing: Conversion Rate ( Physical Event > S2 ),Marketing: Conversion Rate ( Request > S2 ),Marketing: Conversion Rate ( Trial > S2 ),Marketing: MQL Volume,Marketing: MQL volume ( Contact ),Marketing: MQL volume ( Content ),Marketing: MQL volume ( Events ),Marketing: MQL volume ( Trials ),"

    # Create list by splitting comma and removing empty strings
    target_metrics = [x.strip() for x in raw_metrics_string.split(',') if x.strip()]

    processed_data = []

    for metric in target_metrics:
        value = "N/A"

        # CVR Mappings
        if "Content > S2" in metric:
            value = source_data["Content"]["CVR_S2"]
        elif "Digital Event > S2" in metric:
            value = source_data["Digital Event"]["CVR_S2"]
        elif "Physical Event > S2" in metric:
            value = source_data["Physical Event"]["CVR_S2"]
        elif "Request > S2" in metric:
            value = source_data["Request Contact"]["CVR_S2"]
        elif "Trial > S2" in metric:
            value = source_data["Trial"]["CVR_S2"]

        # Volume Mappings
        elif metric == "Marketing: MQL Volume":
            value = source_data["Grand Total"]["MQLs"]
        elif "( Contact )" in metric:
            value = source_data["Request Contact"]["MQLs"]
        elif "( Content )" in metric:
            value = source_data["Content"]["MQLs"]
        elif "( Trials )" in metric:
            value = source_data["Trial"]["MQLs"]
        elif "( Events )" in metric:
            value = source_data["Digital Event"]["MQLs"] + source_data["Physical Event"]["MQLs"]

        processed_data.append({"Metric Name": metric, column_name: value})

    return pd.DataFrame(processed_data).set_index("Metric Name")

# ---------------------------------------------------------
# 3. Apply Function & Join
# ---------------------------------------------------------

df_apac = process_metrics(Apac_Ocr, "APAC")
df_emea = process_metrics(Emea_Ocr, "EMEA")
df_digital = process_metrics(Dig_Ocr, "Digital")
df_smb = process_metrics(Smb_Ocr, "SMB")
df_latam = process_metrics(Latam_Ocr, "LATAM")
df_na = process_metrics(Amer_Ocr, "NA")

# Joining in the specific order requested
final_df = df_apac.join([df_emea, df_digital, df_smb, df_latam, df_na])

# Display
print(final_df.to_markdown())
```
